````python
# Temperature 30 se zyada
df[df.temperature > 30]

# Event Rain ho
df[df.event == "Rain"]

# Windspeed 5 se kam ho
df[df.windspeed < 5]

# Temperature 32 ho
df[df.temperature == 32]

# IMPORTANT:
# Ye 4 alag-alag filters hain, ek hi filter nahi.
# Har line independently DataFrame ko filter karti hai.
# Jupyter Notebook by default sirf LAST line ka output dikhata hai.
# Isliye upar wali conditions (jaise windspeed < 5) last output par apply nahi hoti hain.
# Agar multiple conditions ko ek saath apply karna ho, to '&' (AND) ya '|' (OR) use karna hoga.
# Real projects me agar question me multiple requirements di gayi hon, to combined filtering use karna best practice hai.
````

````python
# IMPORTANT:
# Sabhi conditions ek saath apply ho rahi hain.
# '&' ka matlab AND hai, yani har condition True honi chahiye.
# Agar ek bhi condition False hui, to wo row output me nahi aayegi.
# Ye approach tab use hoti hai jab question me multiple requirements di gayi hon.
# Real-world data analysis me isi tarah combined filtering use ki jati hai.
# Multiple conditions ko ek hi result me accurately filter karne ka ye standard tareeka hai.


df[
    (df.temperature > 30) &      # Temperature 30 se zyada ho
    (df.event == "Rain") &       # Event Rain ho
    (df.windspeed < 5) &         # Windspeed 5 se kam ho
    (df.temperature == 32)       # Temperature exactly 32 ho
]
````

# `DataFrame` Documentation

## Purpose

**DataFrame** Pandas ka ek **2-D (Two-Dimensional)** data structure hai.

- Rows aur Columns me data store karta hai.
- Excel Table ya SQL Table ki tarah dikhta hai.
- Har column ka alag data type ho sakta hai.

---

# Syntax

```python
pd.DataFrame(data, index=None, columns=None, dtype=None, copy=False)
```

---

# Parameters

### `data`

DataFrame banane ke liye input data.

```python
Dictionary
List
Tuple
NumPy Array
Series
CSV
Excel
SQL
```

---

### `index`

Custom Row Index.

```python
index=[101,102,103]
```

---

### `columns`

Column names.

```python
columns=["Name","Age"]
```

---

### `dtype`

Sabhi columns ka data type set karta hai.

```python
dtype=float
```

---

### `copy`

Original data ki copy banani hai ya nahi.

```python
copy=True
```

---

# Examples

## Example 1 (Dictionary)

```python
import pandas as pd

df = pd.DataFrame({
    "Name":["Nishar","Ali","Rahul"],
    "Age":[24,25,26]
})

df
```

---

## Example 2 (List of Lists)

```python
data=[
    ["Nishar",24],
    ["Ali",25],
    ["Rahul",26]
]

df=pd.DataFrame(data,columns=["Name","Age"])
```

---

## Example 3 (NumPy Array)

```python
import numpy as np

arr=np.array([
    [1,2],
    [3,4]
])

df=pd.DataFrame(arr,columns=["A","B"])
```

# Interview Questions & Answers

### Q1. DataFrame kya hai?

**Answer:**  
DataFrame Pandas ka 2-D tabular data structure hai jisme data rows aur columns ke form me store hota hai.

---

### Q2. DataFrame aur Series me difference?

| DataFrame | Series |
|------------|---------|
| 2-D | 1-D |
| Rows aur Columns | Sirf ek Column |
| Multiple Data Types | Single Data Type (generally) |

---

### Q3. DataFrame ka constructor kya hai?

**Answer:**

```python
pd.DataFrame()
```

---

### Q4. DataFrame ka dimension kitna hota hai?

**Answer:**  
2-Dimensional.

---

### Q5. DataFrame me different data types ho sakte hain?

**Answer:**  
Haan. Ek hi DataFrame me Integer, Float, String, Boolean, DateTime sab ho sakte hain.

---

# Interview Important Points

- `DataFrame` Pandas ka **2-D tabular data structure** hai.
- Rows aur Columns me data store karta hai.
- Excel Sheet aur SQL Table ki tarah dikhta hai.
- `pd.DataFrame()` se create kiya jata hai.
- Har column ka data type alag ho sakta hai.
- DataFrame mutable hota hai (data modify kiya ja sakta hai).
- `shape`, `columns`, `index`, `dtypes`, `info()`, `describe()` bahut important properties/methods hain.
- Return type: **DataFrame**.
