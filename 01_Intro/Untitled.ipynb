# Pandas Documentation

## What is Pandas?

**Pandas** Python ki ek **Open Source Data Analysis & Data Manipulation Library** hai.

Iska use structured data (Rows & Columns) ko read, clean, analyze, manipulate aur export karne ke liye kiya jata hai.

Pandas ka naam **"Panel Data"** se aaya hai.

---

# Why Use Pandas?

- Fast Data Analysis
- Data Cleaning
- Missing Data Handling
- Data Filtering
- Sorting
- Grouping
- Merging
- Reshaping
- Reading & Writing Files (CSV, Excel, SQL)
- Data Visualization Support

# Interview Important Points

- **Pandas** Python ki Open Source Data Analysis Library hai.
- Pandas ka naam **Panel Data** se liya gaya hai.
- Alias **`pd`** hota hai.
- Do main Data Structures:
  - **Series (1-D)**
  - **DataFrame (2-D)**
- CSV → `pd.read_csv()`
- Excel → `pd.read_excel()`
- SQL → `pd.read_sql()`
- DataFrame Export → `to_csv()`, `to_excel()`
- Missing Data → `fillna()`, `dropna()`, `interpolate()`
- Combining Data → `merge()`, `concat()`, `join()`
- Reshaping → `pivot()`, `pivot_table()`, `melt()`, `stack()`, `unstack()`
- Grouping → `groupby()`, `crosstab()`
- Visualization → `plot()`
- Pandas NumPy ke upar build ki gayi library hai.
- Data Analysis, Data Science aur Machine Learning me sabse zyada use hone wali libraries me se ek hai.
