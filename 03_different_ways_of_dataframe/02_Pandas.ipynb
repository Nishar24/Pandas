## Different Ways Of Creating Dataframe

## Using csv

````python
import pandas as pd
df = pd.read_csv("weather_data.csv")
df
````

## Using Excel

````python
df = pd.read_excel("weather_data.xlsx")
df
````

## Using Dictionary

````python
weather_data = {
    'day': ['1/1/2017','1/2/2017','1/3/2017'],
    'temperature': [32,35,28],
    'windspeed': [6,7,2],
    'event': ['Rain', 'Sunny', 'Snow']
}

df = pd.DataFrame(weather_data)
df
````

## Using Tuples

````python
weather_data = [
    ('01/01/2026', 32, 6, 'Rain'),   # Day, Temperature, Windspeed, Event
    ('01/02/2026', 25, 7, 'Sunny'),  # Day, Temperature, Windspeed, Event
    ('01/03/2026', 28, 2, 'Snow')    # Day, Temperature, Windspeed, Event
]

# List ko DataFrame me convert karna aur column names dena
df = pd.DataFrame(
    weather_data,
    columns=["day", "temperature", "windspeed", "event"]
)

df   # DataFrame ko display karta hai
````

## Using list of dictionaries

````python
weather_data = [
    {'day': '1/1/2017', 'temperature': 32, 'windspeed': 6, 'event': 'Rain'},
    {'day': '1/2/2017', 'temperature': 35, 'windspeed': 7, 'event': 'Sunny'},
    {'day': '1/3/2017', 'temperature': 28, 'windspeed': 2, 'event': 'Snow'},
    
]

df = pd.DataFrame(data=weather_data, columns=['day','temperature','windspeed','event'])
df
````

# Different Ways to Create a DataFrame

## Purpose

Pandas DataFrame ko kai tarikon se create kiya ja sakta hai.

---

# 1. Using Dictionary

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Nishar", "Ali", "Rahul"],
    "Age": [25, 30, 28]
})

df
```

---

# 2. Using List of Dictionaries

```python
df = pd.DataFrame([
    {"Name": "Nishar", "Age": 25},
    {"Name": "Ali", "Age": 30},
    {"Name": "Rahul", "Age": 28}
])

df
```

---

# 3. Using Dictionary of Lists

```python
data = {
    "Name": ["Nishar", "Ali", "Rahul"],
    "Age": [25, 30, 28]
}

df = pd.DataFrame(data)

df
```

---

# 4. Using List of Tuples

```python
data = [
    ("Nishar", 25),
    ("Ali", 30),
    ("Rahul", 28)
]

df = pd.DataFrame(data,
                  columns=["Name", "Age"])

df
```

---

# 5. Using List of Lists

```python
data = [
    ["Nishar", 25],
    ["Ali", 30],
    ["Rahul", 28]
]

df = pd.DataFrame(data,
                  columns=["Name", "Age"])

df
```

---

# 6. Reading CSV File

```python
df = pd.read_csv("data.csv")

df
```

---

# 7. Reading Excel File

```python
df = pd.read_excel("data.xlsx")

df
```

---

# 8. Using Series

```python
name = pd.Series(["Nishar", "Ali", "Rahul"])
age = pd.Series([25, 30, 28])

df = pd.DataFrame({
    "Name": name,
    "Age": age
})

df
```

---

# 9. Using NumPy Array

```python
import numpy as np

arr = np.array([
    ["Nishar", 25],
    ["Ali", 30],
    ["Rahul", 28]
])

df = pd.DataFrame(arr,
                  columns=["Name", "Age"])

df
```

# Interview Questions & Answers

**Answer:**  
Dictionary aur Dictionary of Lists.

### Q2. CSV file se DataFrame kaise banate hain?

**Answer:**

```python

```

### Q3. Excel file se DataFrame kaise banate hain?


**Answer:**  
Haan, `pd.DataFrame()` me List of Dictionaries pass karke.

**Answer:**

# Interview Important Points

- `pd.DataFrame()` Pandas Library Function hai.
- Dictionary se DataFrame banana sabse common method hai.
- List of Dictionaries API data ke liye useful hai.
- List of Tuples aur List of Lists bhi use hote hain.
- CSV file → `pd.read_csv()`
- Excel file → `pd.read_excel()`
- SQL Database → `pd.read_sql()`
- NumPy Array aur Series se bhi DataFrame create kiya ja sakta hai.
- DataFrame Pandas ka 2-D tabular data structure hai.
